```python
from scipy import stats
import pandas as pd
import numpy as np

# Load data
pv18 = pd.read_csv("/kaggle/input/datasets/belovedorange/research-methodology-project/node_18_PV_actual.csv").drop(columns=["timestamp"])
pv33 = pd.read_csv("/kaggle/input/datasets/belovedorange/research-methodology-project/node_33_PV_actual.csv").drop(columns=["timestamp"])
wind22 = pd.read_csv("/kaggle/input/datasets/belovedorange/research-methodology-project/node_22_wind_actual.csv").drop(columns=["timestamp"])
wind25 = pd.read_csv("/kaggle/input/datasets/belovedorange/research-methodology-project/node_25_wind_actual.csv").drop(columns=["timestamp"])

# Flatten
pv18_vals = pv18.values.flatten()
pv33_vals = pv33.values.flatten()
wind22_vals = wind22.values.flatten()
wind25_vals = wind25.values.flatten()

# Remove NaN
pv18_vals = pv18_vals[~np.isnan(pv18_vals)]
pv33_vals = pv33_vals[~np.isnan(pv33_vals)]
wind22_vals = wind22_vals[~np.isnan(wind22_vals)]
wind25_vals = wind25_vals[~np.isnan(wind25_vals)]

# ANOVA
f_stat, p_val = stats.f_oneway(pv18_vals, pv33_vals, wind22_vals, wind25_vals)

print("F-statistic:", f_stat)
print("p-value:", p_val)

if p_val < 0.05:
    print("Reject Null Hypothesis (H0)")
else:
    print("Accept Null Hypothesis (H0)")
```
